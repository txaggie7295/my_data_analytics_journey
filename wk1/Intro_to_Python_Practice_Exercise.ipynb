## Task 1: Variable Declaration and Data Type

```python
a = "hello"
b = 100
c = 60.50
d = False

print(a, type (a)) 
print(b, type (b))
print(c, type (c))
print(d, type (d))
```

## Task 2: String Manipulation

```python
X_Str = "hello world"

print(X_Str.upper()) # convert to uppercase
print(X_Str.capitalize()) # capitalize first word
print(X_Str[0:5].capitalize()) # parse first word and capitalize
print(X_Str * 3) # display message 3 times
```

## Task 3: Basic Operations

```python
x, y = 20, 5

# Arithmetic
print("Arithmetic:")
print(x + y) 
print(x - y) 
print(x * y) 
print(int(x / y))
print()

# Camparison
print("Comparison:")
print(x > y)
print(x < y)
print(x == y)
print(x != y)
print()

# Logical
print("Logical:")
print(x == 30 and y == 5)
print(x > 5 and y < 20)
print(x < 10 or y > 10) 
print(x < 50 or y > 0)
print(x > y)
print(y > x)
```
