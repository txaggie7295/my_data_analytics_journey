## Task 3: "if-elif-else" Statements

```python
def convert_grade(grade_num):
    if grade_num >= 90:
        return("A")
    elif grade_num >= 80:
        return("B")
    elif grade_num >= 70:
        return("C")
    elif grade_num >= 60:
        return("D")
    else:
        return("F")
```

```python
print("Task 3")
print(f"A number grade of 95 will be an {convert_grade(95)}.") 
print(f"A number grade of 89 will be a {convert_grade(89)}.") 
print(f"A number grade of 71 will be a {convert_grade(71)}.") 
print(f"A number grade of 67 will be a {convert_grade(67)}.") 
print(f"A number grade of 36 will be an {convert_grade(36)}.") 
# convert_grade(95)
# convert_grade(82)
# convert_grade(79)
# convert_grade(65)
# convert_grade(55)
```

## Task 4: Nested Conditionals

```python
def user_access(login, admin, premium):
    if login:
        if admin:
            print("Welcome admin!  Full access is granted")
        elif premium:
            print("Welcome premium user!  Premium content access granted")
        else:
            print("Welcome!  Basic access granted")
    else:
        print("Please log in to access the site")
```

```python
user_access(False,False,False) 
user_access(True,True,False)
user_access(True,False,True)
user_access(True,False,False)
```

## Task 5: Ternary Operator

```python
def check_even_odd(number):
    if number % 2 == 0:
        print("Even")
    else:
        print("Odd")
```

```python
check_even_odd(6)
check_even_odd(5)
```

## Task 6: Combining Conditions

```python
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
        return(True)
    else:
        return(False)
    
```

```python
print(f"Is 1982 a leap year? {is_leap_year(1982)}")
print(f"Is 1988 a leap year? {is_leap_year(1988)}")
print(f"Is 1900 a leap year? {is_leap_year(1900)}")
print(f"Is 2000 a leap year? {is_leap_year(2000)}")
```

## Best Practices (Refactoring)

```python
def classify_item (item, price, stock):
    item_type = "electronics" if item =="electronics" else "non-electronics"
    price_description = "Affordable" if price < 100 else "Expensive"
    in_stock = "in Stock" if stock == 'in' else "out of stock"
    return f"{price_description} {item_type} {in_stock}"
```

```python
print(f"Item Status: {classify_item("electronics",45,"in")}")
print(f"Item Status: {classify_item("electronics",45,"out")}")
print(f"Item Status: {classify_item("electronics",105,"in")}")
print(f"Item Status: {classify_item("electronics",105,"out")}")
print(f"Item Status: {classify_item("plumbing supplies",45,"in")}")
print(f"Item Status: {classify_item("plumbing supplies",45,"out")}")
print(f"Item Status: {classify_item("plumbing supplies",105,"in")}")
print(f"Item Status: {classify_item("plumbing supplies",105,"out")}")

```
