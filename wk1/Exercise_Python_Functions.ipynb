## Task 6: Function Composition

```python
def cels_to_far(celsius):
    return celsius * 9/5 +32
def far_to_cels(farenheit):
    return ((farenheit - 32) * 5/9)
```

```python
print("Celsius to farenheit:")
print(f"0 degress Celsius is equal to {int(cels_to_far(0))} degrees farenheit.")
print(f"100 degress celsius is equal to {int(cels_to_far(100))} degrees farenheit.")
print()
print("Farenheit to celsius:")
print(f"32 degress farenheit is equal to {int(far_to_cels(32))} degrees celsius.")
print(f"212 degress farenheit is equal to {int(far_to_cels(212))} degrees celsius.")
```

## Task 7: Combining Functions and Conditionals

```python
def temp_feel (temp,unit):
    if unit == "F":
        temp = far_to_cels(temp)

    if temp < 0:
        return("freezing")
    elif temp < 15:
        return("cold")
    elif temp < 25:
        return("moderate")
    else:
        return("hot")
```

```python
print(f"30 F is {int(far_to_cels(30))} C and {temp_feel(30,"F")}.")
print(f"40 F is  {int(far_to_cels(40))} C and {temp_feel(40,"F")}.")
print(f"70 F is {int(far_to_cels(70))} C and {temp_feel(70,"F")}.")
print(f"95 F is {int(far_to_cels(95))} C and {temp_feel(95,"F")}.")
```

## Task 8: Function with Variable Arguments

```python
def calc_avg(*numbers):
    return (sum(numbers) / len(numbers))
```

```python
print(int(calc_avg(10,20,30,40,50)))
print(int(calc_avg(12,56,3,98,256,5)))
```

## Task 9: Advanced Challenge - Student Grade Function

```python
def calc_avg_grade(name, *grades):
    if not grades:
        return f"{name} has no scores."
        
    avg_grade = int(sum(grades) / len(grades))

    if avg_grade >= 90:
        grade = "A"
    elif avg_grade >= 80:
        grade = "B"
    elif avg_grade >= 70:
        grade = "C"
    elif avg_grade >= 60:
        grade = "D"
    else:
        grade = "F"

    return f"With an average score of {avg_grade}, {name}'s grade is: {grade}."
```

```python
print(calc_avg_grade("Jeff",99,96,79,90))
print(calc_avg_grade("Lisa",78,92,80,91))
print(calc_avg_grade("David",54,32,65,80),"ha ha!")
```
